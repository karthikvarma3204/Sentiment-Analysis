# Sentiment Analysis Using NLP


## Task Overview
This notebook performs sentiment analysis on textual data using NLP techniques.
It includes data preprocessing, model implementation, and insights.


```

import pandas as pd
import numpy as np
import re
import string
import nltk
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

nltk.download("stopwords")
```

```

# Load dataset
df = pd.read_csv("sentiment_dataset.csv")

# Display first few rows
df.head()
```

```

# Function to clean text
def clean_text(text):
    text = text.lower()  # Convert to lowercase
    text = re.sub(f"[{string.punctuation}]", "", text)  # Remove punctuation
    words = text.split()
    words = [word for word in words if word not in stopwords.words("english")]  # Remove stopwords
    return " ".join(words)

# Apply text cleaning
df["Cleaned_Text"] = df["Text"].apply(clean_text)

df.head()
```

```

# Convert text into numerical features
vectorizer = CountVectorizer()
X_counts = vectorizer.fit_transform(df["Cleaned_Text"])

tfidf_transformer = TfidfTransformer()
X_tfidf = tfidf_transformer.fit_transform(X_counts)

# Convert Sentiments to Numeric Labels
df["Sentiment_Label"] = df["Sentiment"].map({"Positive": 1, "Neutral": 0, "Negative": -1})

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(X_tfidf, df["Sentiment_Label"], test_size=0.2, random_state=42)
```

```

# Train a Naive Bayes sentiment classification model
model = MultinomialNB()
model.fit(X_train, y_train)
```

```

# Make predictions
y_pred = model.predict(X_test)

# Model evaluation
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy}")
print(report)

# Save results
with open("sentiment_results.txt", "w") as f:
    f.write(f"Accuracy: {accuracy}\n")
    f.write(report)
```
